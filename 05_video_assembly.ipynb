```python
import os
import moviepy
import numpy as np 
```

```python
#Load de video as as subclip 

video = moviepy.VideoFileClip(os.getenv("VIDEO_PATH", "data/match.mp4")).subclipped(360,420).without_audio() 
```

```python
#load audio files
audio1 = moviepy.AudioFileClip("data/out_001.wav")
audio2 = moviepy.AudioFileClip("data/out_002.wav")
audio3 = moviepy.AudioFileClip("data/out_003.wav")
audio4 = moviepy.AudioFileClip("data/out_004.wav")
audio5 = moviepy.AudioFileClip("data/out_005.wav")
audio6 = moviepy.AudioFileClip("data/out_006.wav")
concat = moviepy.concatenate_audioclips([audio1, audio2, audio3,audio4, audio5, audio6])
```

```python
video = video.with_audio(concat)
```

```python
video.write_videofile("data/salida.mp4", codec="libx264", audio_codec="aac")
```

```python
video.display_in_notebook()
```
