```python
import os
from omnivoice import OmniVoice
import soundfile as sf
import torch
import polars as pl 
```

https://github.com/k2-fsa/OmniVoice

```python
model = OmniVoice.from_pretrained(
    "k2-fsa/OmniVoice",
    device_map="mps",
    dtype=torch.float16
)
# Apple Silicon users: use device_map="mps" instead
```

```python
Narration = pl.read_csv("data/Narration_1.csv")[:,0].to_list()
```

```python
def generate_audio(frases_narrador):
    for i, text in enumerate(frases_narrador, 1):
        audio = model.generate(
        text= text,
        ref_audio = os.environ["REF_AUDIO"],
        ref_text = os.environ["REF_TEXT"],
        num_steps = 32 
    )
        sf.write(f"data/out_{i:03}.wav", audio[0], 24000)
```

```python
generate_audio(Narration)
```
